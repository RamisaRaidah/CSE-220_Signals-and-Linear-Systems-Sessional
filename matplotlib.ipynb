```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.style.use('default')
```

```python
x=np.linspace(0,10,100)
y=np.sin(x)

plt.figure(figsize=(6,4))
plt.plot(x,y)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Simple Sine Wave')
plt.grid(True)
plt.show()
```

```python
x=np.linspace(0,10,100)
y1=np.sin(x)
y2=np.cos(x)

plt.figure(figsize=(6,4))
plt.plot(x,y1, label='sine wave')
plt.plot(x,y2, label='cosine wave')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Sine and Cosine wave')
plt.legend()
plt.grid(True)
plt.show()
```

```python
fig,ax=plt.subplots(figsize=(6,4))
ax.plot(x,y)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_title('Using Axes Object')
plt.grid(True)
plt.show()
```

```python
fig,ax=plt.subplots(figsize=(6,4))
ax.plot(x,y)
ax.set_xlabel('X')
ax.set_ylabel('Y')

fig, (ax1,ax2)=plt.subplots(1,2,figsize=(20,6))
ax1.plot(x,y1)
ax2.plot(x,y2)

plt.show()
```

```python
fig=plt.figure(figsize=(15,10),dpi=100)
fig.suptitle('Figure Title', fontsize=16)

ax1=fig.add_subplot(221)
ax1.plot(x,y)
ax1.set_title('Axes Title')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')

ax2=fig.add_subplot(222)
ax2.plot(x,y)
ax2.set_title('Axes Title')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')

ax3=fig.add_subplot(223)
ax3.plot(x,y)
ax3.set_title('Axes Title')
ax3.set_xlabel('X')
ax3.set_ylabel('Y')

ax1.xaxis.set_major_locator(plt.MaxNLocator(10))
ax1.yaxis.set_major_formatter(plt.FuncFormatter(lambda x,p:f'${x:.2f}'))

# Another way of doing this without the lambda   --- your_function(tick_value, tick_index)
# def dollar_format(x, p):
#     return f'${x:.0f}'

# ax.yaxis.set_major_formatter(plt.FuncFormatter(dollar_format))

ax1.grid(True)
plt.show()
```

#### Line Plot

```python
x=np.linspace(0,10,100)
y=np.sin(x)

plt.figure(figsize=(6,4))
plt.suptitle(f"But it's golden, like daylight")
plt.plot(x,y,linewidth=2, color='gold', linestyle='-',marker='o',markersize=4)
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

#### Scatter Plot

```python
x=np.random.randn(100)
y=np.random.randn(100)
colors=np.random.rand(100)
sizes=1000*np.random.rand(100)

plt.figure(figsize=(6,4))
plt.scatter(x,y,c=colors, s=sizes, alpha=0.5,cmap='viridis')
plt.colorbar()
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Scatter Plot')
plt.show()
```

#### Bar Plot

```python
categories=['A','B','C','D','E']
values=[23,45,56,78,32]
plt.figure(figsize=(6,4))
plt.bar(categories,values,color='skyblue',edgecolor='black')
plt.xlabel('Categories')
plt.ylabel('Values')

plt.title('Bar Plot')
plt.show()

plt.figure(figsize=(6,4))
plt.barh(categories,values,color='skyblue',edgecolor='black')
plt.xlabel('Categories')
plt.ylabel('Values')

plt.title('Horizontal Bar Plot')
plt.show()
```

#### Histogram

```python
data = np.random.normal(100, 15, 1000)

plt.figure(figsize=(6,4))
plt.hist(data, bins=30, color='steelblue', edgecolor='black', alpha=0.7)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram')
plt.show()

# Multiple histograms
data1 = np.random.normal(100, 15, 1000)
data2 = np.random.normal(120, 15, 1000)

plt.figure(figsize=(6,4))
plt.hist(data1, bins=30, alpha=0.5, label='Group 1', color='blue')
plt.hist(data2, bins=30, alpha=0.5, label='Group 2', color='red')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Multiple Histograms')
plt.legend()
plt.show()
```

#### Box Plot

```python
data = [np.random.normal(0, std, 100) for std in range(1, 5)]

plt.figure(figsize=(10, 6))
plt.boxplot(data, label=['Group 1', 'Group 2', 'Group 3', 'Group 4'])
plt.ylabel('Value')
plt.title('Box Plot')
plt.show()
```

#### Pie Chart

```python
sizes = [30, 25, 25, 20]
labels = ['A', 'B', 'C', 'D']
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']

plt.figure(figsize=(8, 8))
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
plt.title('Pie Chart')
plt.show()
```
